## Inheritance in OOPs

```python
#Inheritance basically means take all the variables and function from one class and can be used to the other class 

class Animal:
    def __init__(self,owner):
        self.owner = owner

    def bark(self):
        print(f"The owner is {self.owner}")
class Pet(Animal):
    def __init__(self , animal , owner):
        super().__init__(owner) # This is used to inherete all the variables as well as function from animal class 
        self.animal = animal 

    def selfdriving(self):
        print(f"tesla support selfdriving")
cat = Animal("het")
dog = Pet("dog" , "het")
dog.bark()
dog.selfdriving()
```

## Inheritance Multiple Classes 

```python
class Animal:
    def __init__(self,owner):
        self.owner = owner

    def bark(self):
        print(f"The owner is {self.owner}")
        
class Pet():
    def __init__(self , animal):
        self.animal = animal 

    def selfdriving(self):
        print(f"tesla support selfdriving")

class gg(Animal , Pet):
    def __init__(self ,owner , animal , gg):
        Animal.__init__(self , owner)
        Pet.__init__(self , animal)
        self.gg = gg

    def gg(self):
        print("this message is gg")

a = gg("het" , "dog" , "g")
```

# Polymorphism In Python 

```python
# Polymorphism basically means different classes can have same function name but can behave differently 
class Dog():
    def __init__(self , name):
        self.name = name

    def bark(self):
        return f"The {self.name} barks like bow"

class Cat():
    def __init__(self,name):
        self.name = name
    def bark(self):
        return f"the {self.name} barks like meow"

def speak(animal):
    print(animal.bark())

a = Dog("ro")
b = Cat("co")
speak(a)
```

# Abstraction In python 

```python
# Here like in the main class we just define all the funciton and the functionalities we write in their respective class 
from abc  import ABC , abstractmethod

class Vehicle(ABC):
    def drive(self):
        print("driviing")

    @abstractmethod
    def start_engine(self):
        pass

class Car(Vehicle):
    def start_engine(self):
        print('Car started')
```

# Operators Overlodaing in Python 

```python
# agar koi two different objects and we need to add both than we use operators overloading 

class Vector:
    def __init__(self , x,y):
        self.x = x
        self.y = y

    def __add__(self ,other): # This function provide a other variable
        return Vector(self.x+other.x,self.y+other.y)

    def __repr__(self): # This functions helps us to determine 
        return f"{self.x} , {self.y}"

y1 = Vector(1,2)
y2 = Vector(2,3)
print(y1+y2)
```
