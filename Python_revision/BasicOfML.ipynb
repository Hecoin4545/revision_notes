# Basic Of Machine Learning

## 3 Types Of machine Learning<br>
1) Supervised ML<br>
2) Unsupervised ML<br>
3) Reinforcement Learning<br>

--------------------------------------------------------
1) supervised ML :-<br>
--It is basically used to predict or classify output on the basis of some inputs<br>
-- Supervised ml includes algorithm like Linear Regression , Ridge & lasso , ElasticNet , Logistic Regression , Decision Tree , Random Forest , Adaboost , xgboost

2) Unsupervised Learning:-<br>
--It basically doesn't provide any output it just form clusters/group on the basis of some data<br>
--This includes algorithm like k means,Hierarichal Mean , DBscam clusting <br>

3) Reinforcement Learning:-<br>
--It is just like it tries to learn on its own <br>
