# Regular Expression In Python

## Regular expression is used to find complex pattern in python 

```python
import re
```

```python
pattern = "was"
text = '''Cyclone Dumazile was a strong tropical cyclone in the South-West Indian Ocean that affected Madagascar and Réunion in early March 2018.
Dumazile originated from a cyclone Dyclone low-pressure area that formed near Agaléga on 27 February. It became a tropical disturbance on 2 March,
and was named the next day after attaining tropical storm status. Dumazile reached its peak intensity on 5 March, with 10-minute sustained winds of
165 km/h (105 mph), 1-minute sustained winds of 205 km/h (125 mph), and a central atmospheric pressure of 945 hPa (27.91 inHg). As it tracked 
southeastwards, Dumazile weakened steadily over the next couple of days due to wind shear, and became a post-tropical cyclone on 7 March
'''

match = re.search(pattern , text)  # This will search the first occurance and will tell ki ye hai ya nhi text mai 
print(match)
print(match.span())
```

```python
pattern = r"[A-Z]+yclone" # Here starting mai r use karne se \n jase special character ko bhi string hi consider karege 
print(r"het\n") #output:- het\n

match = re.search(pattern , text) # WIll only give the first occurance

# If we want all the occurance of a particular pattern than 
matches = re.finditer(pattern , text)
for match in matches:
    print(match)
# Agar hume wahi words hi sirf print karvana ho than 
    a = text[match.span()[0]:match.span()[1]]
    print(a)
```

```python
pattern = "was"
text = '''Cyclone Dumazile was a strong tropical cyclone in the South-West Indian Ocean that affected Madagascar and Réunion in early March 2018.
Dumazile originated from a cyclone Dyclone low-pressure area that formed near Agaléga on 27 February. It became a tropical disturbance on 2 March,
and was named the next day after attaining tropical storm status. Dumazile reached its peak intensity on 5 March, with 10-minute sustained winds of
165 km/h (105 mph), 1-minute sustained winds of 205 km/h (125 mph), and a central atmospheric pressure of 945 hPa (27.91 inHg). As it tracked 
southeastwards, Dumazile weakened steadily over the next couple of days due to wind shear, and became a post-tropical cyclone on 7 March
'''

re.match(pattern , text) # It will only check if the first word matches the pattern but if the first word doesnot matches than it won't return anything

a = re.findall(pattern , text) # Finds all occurrences of the pattern in the string and returns them as a list.
print(a)

b = re.sub(pattern ,"a", text) # Replaces occurrences of the pattern with a specified string.
b

# . --> Matches any character except a newline
re.search(r"H.llo", "Hello")  # Matches because '.' can be 'e'

# $: Matches the end of the string.
re.search(r"World$", "Hello World")  # Matches

# []: Matches any one character inside the brackets.
re.search(r"[aeiou]", "Hello")  # Matches 'e'

```

```python
# Character classes
# .	any character except newline
# \w\d\s	word, digit, whitespace
# \W\D\S	not word, digit, whitespace
# [abc]	any of a, b, or c
# [^abc]	not a, b, or c
# [a-g]	character between a & g
# Anchors
# ^abc$	start / end of the string
# \b\B	word, not-word boundary
# Escaped characters
# \.\*\\	escaped special characters
# \t\n\r	tab, linefeed, carriage return
# Groups & Lookaround
# (abc)	capture group
# \1	backreference to group #1
# (?:abc)	non-capturing group
# (?=abc)	positive lookahead
# (?!abc)	negative lookahead
# Quantifiers & Alternation
# a*a+a?	0 or more, 1 or more, 0 or 1
# a{5}a{2,}	exactly five, two or more
# a{1,3}	between one & three
# a+?a{2,}?	match as few as possible
# ab|cd	match ab or cd
```
