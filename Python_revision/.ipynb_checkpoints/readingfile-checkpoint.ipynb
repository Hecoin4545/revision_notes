```python
import re
import json
import pandas as pd
# ─────────────────────────────────────────────────────────────────────────────
# 1) LOAD YOUR EXCEL AND BUILD xlsx_dict
# ─────────────────────────────────────────────────────────────────────────────
xlsx_path = "C:\\Users\\Het\\Downloads\\messages.php"
sheet_name = "Sheet1"
# Columns: “Key” (e.g. "side_menu/dashboard") and “Corrected Telugu Content”
df = pd.read_excel(xlsx_path, sheet_name=sheet_name)
# Drop any rows where either column is NaN
xlsx_data = df[['Key', 'Corrected Telugu Content']].dropna()
# Make a simple dict: { "side_menu/dashboard": "New Telugu string", … }
xlsx_dict = dict(zip(xlsx_data['Key'], xlsx_data['Corrected Telugu Content']))
# ─────────────────────────────────────────────────────────────────────────────
# 2) READ THE PHP FILE AS TEXT AND ISOLATE THE ARRAY LITERAL
# ─────────────────────────────────────────────────────────────────────────────
php_path      = "/Users/imac/Desktop/Translate/i18n/te.php"   # your existing PHP file
output_path   = "/Users/imac/Desktop/Translate/updated_te.php" # where we’ll write the updated version
with open(php_path, 'r', encoding='utf-8') as f:
    php_text = f.read()
# We assume the file looks roughly like:
#    <?php
#
#    return [
#        'side_menu' => [
#            'dashboard' => 'Quick Links',
#            'label_logout' => 'Sign Out',
#            # … many nested keys …
#        ],
#        'notification' => [ … ],
#        # …
#    ];
#
# Possibly with comments, blank lines, etc.
#
# We want to extract everything from the first "[" after “return” down to its matching “];”.
match = re.search(r'return\s*(\[[\s\S]*\]);', php_text)
if not match:
    raise RuntimeError("Could not locate a top‐level PHP array after “return”. Make sure your file indeed has “return [ … ];”.")
php_array_block = match.group(1)  # e.g. the full "[ 'side_menu' => [ ... ], 'notification' => [ ... ], … ]"
# ─────────────────────────────────────────────────────────────────────────────
# 3) CONVERT THE PHP ARRAY SYNTAX INTO JSON SYNTAX
# ─────────────────────────────────────────────────────────────────────────────
# a) Replace “=>” with “:” so that it becomes JSON‐like
json_like = php_array_block.replace("=>", ":")
# b) Replace every single‐quoted string '…' with double quotes. 
#    We assume no nested single quotes inside strings.
#    A quick regex: find '([^']*)'  →  "…"
#    (If some strings contain escaped quotes, you may need a more robust approach.)
def single_to_double_quotes(match):
    inner = match.group(1)
    # Escape any double‐quotes inside inner, if present
    inner_escaped = inner.replace('"', '\\"')
    return f'"{inner_escaped}"'
json_like = re.sub(r"'([^']*)'", single_to_double_quotes, json_like)
# c) Now ensure that the result is valid JSON. 
#    Some PHP arrays allow trailing commas. Python’s json.loads will accept trailing commas in most versions,
#    but if your environment rejects them, you can strip them via regex. Example (remove comma before ] or }):
json_like = re.sub(r",\s*([\]\}])", r"\1", json_like)
# At this point, json_like should be something like:
#    [
#      "side_menu": [
#          "dashboard": "Quick Links",
#          "label_logout": "Sign Out"
#      ],
#      "notification": { … },
#      …
#    ]
#
# Which is valid JSON (a top‐level array/dict). Load it:
php_dict = json.loads(json_like)
# ─────────────────────────────────────────────────────────────────────────────
# 4) UPDATE php_dict IN PLACE ACCORDING TO YOUR Excel DICT
# ─────────────────────────────────────────────────────────────────────────────
def update_dict_with_hierarchical_keys(dct, excel_dict):
    """
    Recursively descend into dct using each hierarchical key (split by . or /),
    and overwrite the final leaf with the Excel value.
    """
    for hierarchical_key, new_value in excel_dict.items():
        # Split on either '.' or '/'
        parts = re.split(r"[./]", hierarchical_key)
        current = dct
        for part in parts[:-1]:
            # If the path doesn’t exist, create a new nested dict
            if part not in current or not isinstance(current[part], dict):
                current[part] = {}
            current = current[part]
        # Overwrite leaf:
        leaf = parts[-1]
        current[leaf] = new_value
    return dct
# Apply updates
updated_dict = update_dict_with_hierarchical_keys(php_dict, xlsx_dict)
# ─────────────────────────────────────────────────────────────────────────────
# 5) CONVERT THE UPDATED Python DICT BACK INTO PHP ARRAY SYNTAX
# ─────────────────────────────────────────────────────────────────────────────
def dict_to_php_array(obj, indent=0):
    """
    Turn a Python dict (possibly nested) into a PHP array literal string.
    - Strings become single-quoted.
    - Keys become single-quoted.
    - Nested dicts become nested brackets.
    - We preserve indentation for readability.
    """
    sp = "    "  # 4 spaces per indent‐level
    pad = sp * indent
    lines = ["["]
    for key, val in obj.items():
        # Prepare the PHP‐style key (always single‐quoted):
        php_key = f"'{key}'"
        if isinstance(val, dict):
            # Recursively format nested dict
            nested = dict_to_php_array(val, indent + 1)
            lines.append(f"{pad}{sp}{php_key} => {nested},")
        else:
            # Leaf node → single-quote and escape any single quotes in val
            escaped_val = str(val).replace("'", "\\'")
            lines.append(f"{pad}{sp}{php_key} => '{escaped_val}',")
    lines.append(pad + "]")
    # Join with newlines, preserving indentation
    return "\n".join(lines)
php_array_new = dict_to_php_array(updated_dict, indent=0)
# Now php_array_new is something like:
#   [
#       'side_menu' => [
#           'dashboard'    => 'Quick Links (overwritten?)',
#           'label_logout' => 'నవీకరించబడిన విలువ',
#           // … etc.
#       ],
#       'notification' => [ … ],
#       // …
#   ]
#
# We need to wrap it back into “return …;” (and keep the “<?php” above).
# Extract everything *before* the original "return [" and everything *after* the matching "];"
prefix = php_text[: match.start(1) ]   # includes “<?php” and whitespace up to the “[”
suffix = php_text[ match.end(1) : ]     # includes “];” and anything after (comments, newline, etc.)
# Now splice in:
new_php_text = prefix + php_array_new + suffix
# ─────────────────────────────────────────────────────────────────────────────
# 6) WRITE THE UPDATED PHP BACK TO DISK
# ─────────────────────────────────────────────────────────────────────────────
with open(output_path, 'w', encoding='utf-8') as f:
    f.write(new_php_text)
print(f"Updated PHP‐array written to: {output_path}")
```
