# Estimates In Statistics 

It is specified obsered numerical value used to estimate an unknown population parameter <br>

1) Point Estimate:- <br>
--single numerical value used to estimate an unknown population parameter <br>
--eg:- sample mean is a point estimate of population mean <br>

2) Interval Estimate:- <br>
-- Range of values used to estimate an unknow population parameter <br>
